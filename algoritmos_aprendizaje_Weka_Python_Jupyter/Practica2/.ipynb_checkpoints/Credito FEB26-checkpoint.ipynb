```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_graphviz
```

```python
tabla = pd.read_csv("./Credito.csv")
tabla
```

```python
for c, a in enumerate(tabla.columns):
    conversion = {v: i for i,v in enumerate(tabla[a].unique())}
    print(conversion)
    tabla.iloc[:,c] = [conversion[tabla.iloc[idx, c]] for idx in tabla.index]
tabla
```

```python
tabla = pd.read_csv("./Credito.csv")
conversion=[{'alto': 2, 'moderado': 1, 'bajo': 0}, 
            {'mala': 0, 'desconocida': 1, 'buena': 2}, 
            {'alta': 1, 'baja': 0}, 
            {'no': 0, 'adecuados': 1}, 
            {'0 a 2M': 0, '2 a 5M': 1, 'más de 5M': 2}]
for c, a in enumerate(tabla.columns):
    tabla.iloc[:,c] = [conversion[c][tabla.iloc[idx, c]] for idx in tabla.index]
tabla
```

```python
clf = DecisionTreeClassifier(criterion='entropy')
```

```python
X = np.array(tabla.iloc[:, 1:], dtype=float).reshape(-1,4)
y = np.array(tabla.iloc[:,0], dtype=int).reshape(-1,1)
```

```python
clf.fit(X, y)
```

```python
export_graphviz(clf, out_file='tree_credito_CART.dot', feature_names=tabla.columns[1:])
```

```python
!dot -Tpng tree_credito_CART.dot -o tree_credito_CART.png -Gdpi=300
```

```python
clf.score(X, y)
```
